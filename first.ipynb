```python
# User se ek number lo aur uske digits ka sum calculate karo.

# Example: 583 → 5 + 8 + 3 = 16

# Conditions:
# - while loop use karna hai.
# - % 10 se last digit nikaalna hai.
# - // 10 se last digit remove karna hai.
# - sum() function use nahi karna.

num = int(input("Enter a 4 digit number"))
if num<1000 :
      print("Please enter 4 or more digit number")
else :
      sum = 0
      while num>0 :         
            last_num = num%10
            sum= sum + last_num
            num = num//10
            
    
print(sum)
```

```python
# Ek program banao:
# 1. Secret number 7 rakho.
# 2. User se number guess karwao.
# 3. Guess 7 se chhota ho toh Too low! print karo.
# 4. Guess 7 se bara ho toh Too high! print karo.
# 5. Sahi guess ho toh Correct guess! print karo aur game stop kar do.
# Extra condition: User ko maximum 3 attempts milne chahiye.
secret_num = 7
attempts = 0

while attempts<3: 
   guessed_num = int(input("Guess a number"))
   attempts += 1
   if guessed_num == secret_num :
     print("Correct Guess!")
     break
   elif guessed_num> secret_num:
    print("Too High!")
   else :
    print("Too low!")

if attempts == 3 and guessed_num != secret_num:
    print("Game Over! The secret number was 7.")
 
```

```python
def square_1(l) :
    square = []
    for i in l :
        square.append(i**2)
    return square

numbers = list(range(1,11))
print(square_1(numbers))
```

```python
def reverse(l) :
    reverse_list=[]
    i = len(l) - 1
    while i>0:
        reverse_list.append(i)
        i-=1
    return reverse_list

numbers = list(range(1,11))
print(reverse(numbers))

```

```python
def reverse(l):
    reversed_list = []
    for i in l:
        reversed_list.append(i[::-1])

    return reversed_list

words = ['abc', 'bchd']
print(reverse(words))
```

```python
def odd_even(l) :
    even = []
    odd = []
    for i in l :
        if i%2==0 :
           even.append(i)
        else :
           odd.append(i)
    return [odd,even]

numbers = list(range(1,50))
print(odd_even(numbers))
```
